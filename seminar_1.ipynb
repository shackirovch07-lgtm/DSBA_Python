```python
import csv
from collections import Counter

path = 'C:/Users/shack/Downloads/dataset.csv/dataset.csv'

seen = set()
feat_cnt = Counter()

with open(path, encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        track_id = row['track_id']
        
        if track_id not in seen:
            seen.add(track_id)
            artists = [a.strip() for a in row['artists'].split(';')]
            
            if len(artists) > 1:
                for artist in artists:
                    feat_cnt[artist] += 1


print("фиты")
for i, (artist, count) in enumerate(feat_cnt.most_common(10), 1):
    print(f"{i}. {artist} — {count} фитов")
```

```python
import csv
from collections import defaultdict

path = 'C:/Users/shack/Downloads/dataset.csv/dataset.csv'

seen = set()
artist_pops = defaultdict(list)

with open(path, encoding='utf-8') as f:
    for row in csv.DictReader(f):
        if row['track_id'] in seen:
            continue
        seen.add(row['track_id'])
        if ';' not in row['artists']:
            try:
                artist = row['artists'].strip()
                artist_pops[artist].append(float(row['popularity']))
            except ValueError:
                pass

top = sorted([(a, sum(p)/len(p)) for a, p in artist_pops.items()], key=lambda x: x[1], reverse=True)[:10]

for i, (artist, avg) in enumerate(top, 1):
    print(f"{i}.{artist} — {avg:.2f}")
```

```python
import csv

path = 'C:/Users/shack/Downloads/dataset.csv/dataset.csv'

with open(path, mode='r', encoding='utf-8') as file:
    reader = csv.reader(file)
    header = next(reader)
    
    artist_index = header.index('artists')
    
    # Словарь для подсчета фитов
    feat_counts = {}

    for row in reader:
        raw_artists = row[artist_index]
        
        # Если в строке есть ';', значит это совместный трек
        if ';' in raw_artists:
            # Разбиваем строку на список отдельных имен
            artists_list = raw_artists.split(';')
            
            for artist in artists_list:
                artist = artist.strip() # убираем лишние пробелы, если они есть
                if artist in feat_counts:
                    feat_counts[artist] += 1
                else:
                    feat_counts[artist] = 1

    # Ищем самого "дружелюбного" артиста
    if feat_counts:
        top_feat_artist = max(feat_counts, key=feat_counts.get)
        count = feat_counts[top_feat_artist]
        
        print(f"--- Король фитов ---")
        print(f"Имя: {top_feat_artist}")
        print(f"Количество совместных треков: {count}")
    else:
        print("Фитов не найдено.")
```

```python
from natasha import (
    Segmenter,
    MorphVocab,
    NewsEmbedding,
    NewsMorphTagger,
    Doc
)

# Инициализация инструментов
segmenter = Segmenter()
morph_vocab = MorphVocab()
emb = NewsEmbedding()
morph_tagger = NewsMorphTagger(emb)

# Тестовый текст
text = 'Natasha — библиотека для обработки естественного языка'
doc = Doc(text)

# Обработка
doc.segment(segmenter)
doc.tag_morph(morph_tagger)

for token in doc.tokens:
    token.lemmatize(morph_vocab)
    print(f'{token.text} -> {token.lemma}')
```
